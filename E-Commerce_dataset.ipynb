## Orders & Transaction Analytics

```python
# 1.Which months generate the highest sales?
highest_sales = sales.groupby('month')['net_sales'].sum().max()
print(f"Highest monthly sales : ${highest_sales:,.2f}")
```

```python
# 2.How does order volume change over time?
order_value = sales.groupby('month')['order_id'].nunique()
plt.bar(order_value.index,order_value.values)

plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.title('Monthly Order Volume')

plt.xticks(rotation=45)
plt.show()
```

```python
#3.Are there seasonal purchasing patterns?
quarterly_trend = sales.groupby('quarter')['order_id'].nunique()
plt.plot(quarterly_trend.index.astype(str),quarterly_trend.values)
plt.xlabel('Quarter')
plt.ylabel('Number of Orders')
plt.title('Quarterly Order Volume')

plt.xticks(rotation=45)
plt.show()
```

```python
#4.Which periods show unusual growth or decline?
monthly_orders = (
    sales.groupby('month')['order_id']
    .nunique()
    .sort_index()
)
monthly_growth = monthly_orders.pct_change() * 100

unsual_growth = monthly_growth[monthly_growth > 20]
unsual_decline = monthly_growth[monthly_growth < -20]

print(f"Unsual growth: {unsual_growth}")
print(f"Unsual Decline: {unsual_decline}")
```

## Customer Analytics & Customer Intelligence

```python
#5.Who are the highest-value customers?
top_customers = sales.groupby('customer_name')['net_sales'].sum().sort_values(ascending=False).head(5)
print(f"Top Customers : {top_customers}")
```

```python
#6.Which customer groups generate the most revenue
sales['age_group'] = pd.cut(
    sales['customer_age'],
    bins=[18, 30, 45, 60, float('inf')],
    labels=['Young Adult', 'Adult', 'Middle-aged', 'Senior']
)

#print(sales[['customer_age', 'age_group']].head())

age_segments_revenue = (
    sales.groupby('age_group')['net_sales']
        .sum()
        .sort_values(ascending=False)
        .head(10)
)
plt.bar(age_segments_revenue.index,age_segments_revenue.values)

plt.xlabel('Age group')
plt.ylabel('Net sales')
plt.title('Revenue by age group')

plt.xticks(rotation=45)
plt.show()
```

```python
#7.Does loyalty relate to purchasing frequency or customer value?
customer_metrics = sales.groupby('customer_id').agg(
    purchase_frequency=('order_id', 'nunique'),
    customer_value=('net_sales', 'sum')
).reset_index() 
   
print(customer_metrics)
```

```python
#8.Do purchasing patterns differ between regions or demographic groups?
region_customer_frequency = (
    sales.groupby(['region', 'customer_id'])['order_id']
      .nunique()
      .groupby('region')
      .mean()
)

print(region_customer_frequency)
```

## Product, Category & Brand Analytics

```python
#9.Product Demand
# Create a demand score (example: rating divided by price to balance affordability and popularity)
products["demand_score"] = products["product_rating"] / products["unit_price"]

# Sort products by demand score
demand_sorted = products.sort_values(by="demand_score", ascending=False)

# Display top 10 products with highest demand
print(demand_sorted[["product_id", "product_name", "product_category", "unit_price", "product_rating", "demand_score"]].head(10))
```

```python
#10.Category performance
#merge orders dataset with products one
order_items_merged = orders.merge(products, on='product_id', how='left')
#Top categories by sales
category_sales = order_items_merged.groupby('product_category')['net_sales'].sum().sort_values(ascending=True)
category_profit = order_items_merged.groupby('product_category')['profit'].sum().sort_values(ascending=True)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

category_sales.plot(kind='barh', ax=axes[0], color='skyblue')
axes[0].set_title('Total Sales by Product Category')
axes[0].set_xlabel('Sales ($)')

category_profit.plot(kind='barh', ax=axes[1], color='orange')
axes[1].set_title('Total Profit by Product Category')
axes[1].set_xlabel('Profit ($)')

plt.tight_layout()
plt.show()

print("\nCategory Performance:")
print(category_sales)
```

```python
# Brand performance
brand_sales = order_items_merged.groupby('brand')['net_sales'].sum().sort_values(ascending=False).head(10)
brand_profit = order_items_merged.groupby('brand')['profit'].sum().sort_values(ascending=False).head(10)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

brand_sales.plot(kind='bar', ax=axes[0], color='purple')
axes[0].set_title('Top 10 Brands by Sales')
axes[0].set_xlabel('Brand')
axes[0].set_ylabel('Net Sales ($)')
axes[0].tick_params(axis='x', rotation=45)

brand_profit.plot(kind='bar', ax=axes[1], color='green')
axes[1].set_title('Top 10 Brands by Profit')
axes[1].set_xlabel('Brand')
axes[1].set_ylabel('Profit ($)')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## Regional Performance

```python
#.regional performance
regional_sales = sales.groupby('region')['net_sales']
                      
#print(regional_sales)
regional_pct = regional_sales/ regional_sales.sum() * 100
#print(regional_pct)

plt.figure(figsize=(5,5))
colors = plt.get_cmap('tab20').colors


plt.pie(regional_pct, labels=regional_pct.index, autopct='%1.1f%%', startangle=90, colors=colors, shadow=True)
plt.title("Purchases By Region")
plt.show()
```

```python
# regions with the highest sales
regional_sales= (
            sales.groupby('region')['net_sales']
            .sum()
           .sort_values(ascending=False)
            .head(10)
)
regional_profit = (
            sales.groupby('region')['profit']
            .sum()
           .sort_values(ascending=False)
            .head(10)
)

#print(regional_performance)
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

regional_sales.plot(kind='bar', ax=axes[0], color='red')
axes[0].set_title('Top 10 Regions By Sales')
axes[0].set_xlabel('Region')
axes[0].set_ylabel('Net Sales ($)')
axes[0].tick_params(axis='x', rotation=45)

regional_profit.plot(kind='bar', ax=axes[1], color='blue')
axes[1].set_title('Top 10 Regions By Profit')
axes[1].set_xlabel('Region')
axes[1].set_ylabel('Net Profit ($)')
axes[1].tick_params(axis='x', rotation=45)



plt.tight_layout()
plt.show()
```
